```python
3. Update

3.1 Dropping Rows
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
3.2 Dropping Columns
```

```python
df.drop('TotalCharges', axis=1).head()
```

```python
3.3 Creating Calculated Columns
```

```python
df['New Column'] = df['MonthlyCharges'] + df['tenure']
df.head()
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['New Column'] = df['MonthlyCharges'] + df['TotalCharges']
df.head()
```

3.5 Updating a Sinle Value

```python
df.loc[0, 'tenure'] = 100
```

```python
df.loc[0:2, ['tenure', 'MonthlyCharges']]
```

```python
3.6 Condition Based Upading Using Apply
```

```python
df['ChargeLevel'] = df['MonthlyCharges'].apply(lambda x: 'High' if x > 70 else 'Low')
```

```python
df[['MonthlyCharges','ChargeLevel']].head()
```

```python
df['Risk'] = df.apply(lambda row: 'High Risk' if row['Contract'] == 'Month-to-month' and row['tenure'] < 12 else 'Low Risk', axis=1)
```

```python
df.head()
```
